```python
# ================================================================
# QUANTUM-ENHANCED HYPERGRAPH WEB USAGE MINING
# FOR EXPLAINABLE LEARNER BEHAVIOR ANALYTICS
#
# COMPLETE PIPELINE
#
# 1. Learner Interaction Data Collection
# 2. Data Preprocessing + Behavioral Feature Construction
# 3. Quantum-Enhanced Behavioral Representation
#       QFE -> trained VQC
# 4. Temporal Learner Behavior Modeling
#       behavioral-state sequence
# 5. Higher-Order Relational Modeling
#       similarity hypergraph + DHGNN-style representation
# 6. Learning-Style / Target Prediction
#       QSVM / quadratic SVM
# 7. Explainable Learner Behavior Analytics
#       SHAP + LIME
# 8. Actionable Counterfactual Recommendation
#       APO-style counterfactual optimization
#
# INPUT:
#   online_classroom_data.csv
#
# CURRENT DATASET:
#   71 rows x 16 columns
#
# IMPORTANT:
#   The current CSV contains "Approved" as the available target.
#   It does NOT contain timestamp/session fields or four explicit
#   FLSM ground-truth labels. The code therefore does not invent them.
# ================================================================


# ================================================================
# STEP 0 - INSTALL REQUIRED PACKAGES
# ================================================================

!pip -q install pennylane scikit-learn pandas numpy matplotlib seaborn openpyxl shap lime


# ================================================================
# STEP 1 - IMPORT LIBRARIES
# ================================================================

import os
import re
import random
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
    roc_curve
)
from sklearn.cluster import SpectralClustering
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.neighbors import NearestNeighbors

import pennylane as qml

import torch
import torch.nn as nn
import torch.optim as optim

warnings.filterwarnings("ignore")

SEED = 42

np.random.seed(SEED)
random.seed(SEED)
torch.manual_seed(SEED)

print("Libraries loaded successfully.")


print("\nUpload your online_classroom_data.csv file if it is not already present.")

possible_paths = [
    "/content/drive/MyDrive/online_classroom_data.csv"
]

INPUT_FILE = None

for path in possible_paths:
    if os.path.exists(path):
        INPUT_FILE = path
        break

if INPUT_FILE is None:
    uploaded = files.upload()

    uploaded_names = list(uploaded.keys())

    csv_files = [
        f for f in uploaded_names
        if f.lower().endswith(".csv")
    ]

    if len(csv_files) == 0:
        raise FileNotFoundError(
            "No CSV file was uploaded."
        )

    INPUT_FILE = os.path.join("/content", csv_files[0])

print("Input file:", INPUT_FILE)


# ================================================================
# STEP 3 - LOAD DATA
# ================================================================

df_raw = pd.read_csv(INPUT_FILE)

print("\n================ RAW DATA =================")
print("Shape:", df_raw.shape)
print("\nColumns:")
print(df_raw.columns.tolist())

print("\nFirst 5 rows:")
display(df_raw.head())


# ================================================================
# STAGE 1
# LEARNER INTERACTION DATA COLLECTION
# ================================================================

print("\n")
print("=" * 70)
print("STAGE 1 - LEARNER INTERACTION DATA COLLECTION")
print("=" * 70)

print("""
The dataset represents learner interaction information collected
from an e-learning classroom environment.

Available interaction variables include:
- total_posts
- helpful_post
- nice_code_post
- collaborative_post
- confused_post
- creative_post
- bad_post
- amazing_post
- timeonline
- classroom skill measurements
- Approved
""")

# Remove accidental index column
index_like_columns = [
    col for col in df_raw.columns
    if str(col).lower().startswith("unnamed")
]

df = df_raw.drop(columns=index_like_columns, errors="ignore").copy()

# Remove duplicate rows
before_duplicates = len(df)

df = df.drop_duplicates().reset_index(drop=True)

after_duplicates = len(df)

print(
    f"Duplicate rows removed: "
    f"{before_duplicates - after_duplicates}"
)

print("Current dataset shape:", df.shape)


# ================================================================
# STAGE 2
# DATA PREPROCESSING
# ================================================================

print("\n")
print("=" * 70)
print("STAGE 2 - DATA PREPROCESSING")
print("=" * 70)


# ------------------------------------------------
# 2.1 Convert numeric columns
# ------------------------------------------------

numeric_columns = [
    "total_posts",
    "helpful_post",
    "nice_code_post",
    "collaborative_post",
    "confused_post",
    "creative_post",
    "bad_post",
    "amazing_post",
    "timeonline",
    "Approved"
]

skill_columns = [
    "sk1_classroom",
    "sk2_classroom",
    "sk5_classroom",
    "sk3_classroom",
    "sk4_classroom"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )


# ------------------------------------------------
# 2.2 Convert classroom skill values
#
# Some cells contain values such as:
# "2,1"
# "5,7"
#
# These are converted into their average.
# ------------------------------------------------

def parse_skill_value(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if value == "":
        return np.nan

    # Already numeric
    try:
        return float(value)
    except:
        pass

    # Values such as "2,1"
    parts = re.findall(
        r"-?\d+(?:\.\d+)?",
        value
    )

    if len(parts) == 0:
        return np.nan

    numbers = [float(x) for x in parts]

    return np.mean(numbers)


for col in skill_columns:

    if col in df.columns:

        df[col + "_Numeric"] = df[col].apply(
            parse_skill_value
        )


# ------------------------------------------------
# 2.3 Handle invalid values
# ------------------------------------------------

behavior_raw_columns = [
    "total_posts",
    "helpful_post",
    "nice_code_post",
    "collaborative_post",
    "confused_post",
    "creative_post",
    "bad_post",
    "amazing_post",
    "timeonline"
]

for col in behavior_raw_columns:

    if col in df.columns:

        df.loc[
            df[col] < 0,
            col
        ] = np.nan


# ------------------------------------------------
# 2.4 Missing value imputation
# ------------------------------------------------

numeric_df_columns = df.select_dtypes(
    include=np.number
).columns

for col in numeric_df_columns:

    if col == "Approved":
        continue

    median_value = df[col].median()

    if pd.isna(median_value):
        median_value = 0

    df[col] = df[col].fillna(
        median_value
    )


print("\nMissing values after preprocessing:")

print(
    df.isnull().sum().sum()
)


# ================================================================
# STAGE 2B
# BEHAVIORAL FEATURE CONSTRUCTION
# ================================================================

print("\n")
print("=" * 70)
print("BEHAVIORAL FEATURE CONSTRUCTION")
print("=" * 70)


# ------------------------------------------------
# Prevent division by zero
# ------------------------------------------------

posts = df["total_posts"].replace(
    0,
    np.nan
)


# ------------------------------------------------
# Core behavioral features
# ------------------------------------------------

df["Total_Interaction_Count"] = (
    df[
        [
            "total_posts",
            "helpful_post",
            "nice_code_post",
            "collaborative_post",
            "confused_post",
            "creative_post",
            "bad_post",
            "amazing_post"
        ]
    ].sum(axis=1)
)


df["Average_Post_Activity"] = (
    df["Total_Interaction_Count"] / 7
)


df["Helpful_Post_Rate"] = (
    df["helpful_post"] / posts
)


df["Nice_Code_Post_Rate"] = (
    df["nice_code_post"] / posts
)


df["Collaborative_Post_Rate"] = (
    df["collaborative_post"] / posts
)


df["Confusion_Post_Rate"] = (
    df["confused_post"] / posts
)


df["Creative_Post_Rate"] = (
    df["creative_post"] / posts
)


df["Negative_Post_Rate"] = (
    df["bad_post"] / posts
)


df["Amazing_Post_Rate"] = (
    df["amazing_post"] / posts
)


df["Positive_Engagement_Count"] = (
    df["helpful_post"]
    + df["nice_code_post"]
    + df["collaborative_post"]
    + df["creative_post"]
    + df["amazing_post"]
)


df["Positive_Engagement_Rate"] = (
    df["Positive_Engagement_Count"] / posts
)


df["Help_Seeking_Negative_Count"] = (
    df["helpful_post"]
    + df["confused_post"]
)


df["Help_Seeking_Negative_Rate"] = (
    df["Help_Seeking_Negative_Count"] / posts
)


df["Time_Online_Minutes"] = (
    df["timeonline"]
)


df["Time_Per_Post"] = (
    df["timeonline"] / posts
)


df["Interaction_Density"] = (
    df["Total_Interaction_Count"]
    / (df["timeonline"] + 1)
)


# ------------------------------------------------
# Classroom skill statistics
# ------------------------------------------------

numeric_skill_columns = [
    col + "_Numeric"
    for col in skill_columns
]

df["Classroom_Skill_Mean"] = (
    df[numeric_skill_columns].mean(axis=1)
)

df["Classroom_Skill_Std"] = (
    df[numeric_skill_columns].std(axis=1)
)

df["Classroom_Skill_Max"] = (
    df[numeric_skill_columns].max(axis=1)
)

df["Classroom_Skill_Min"] = (
    df[numeric_skill_columns].min(axis=1)
)

df["Classroom_Skill_Range"] = (
    df["Classroom_Skill_Max"]
    - df["Classroom_Skill_Min"]
)


# ------------------------------------------------
# Replace infinity
# ------------------------------------------------

df = df.replace(
    [np.inf, -np.inf],
    np.nan
)


# ------------------------------------------------
# Final imputation
# ------------------------------------------------

for col in df.select_dtypes(
    include=np.number
).columns:

    if col == "Approved":
        continue

    df[col] = df[col].fillna(
        df[col].median()
    )


# ================================================================
# BEHAVIORAL FEATURE LIST
# ================================================================

behavioral_features = [
    "Total_Interaction_Count",
    "Average_Post_Activity",
    "Helpful_Post_Rate",
    "Nice_Code_Post_Rate",
    "Collaborative_Post_Rate",
    "Confusion_Post_Rate",
    "Creative_Post_Rate",
    "Negative_Post_Rate",
    "Amazing_Post_Rate",
    "Positive_Engagement_Count",
    "Positive_Engagement_Rate",
    "Help_Seeking_Negative_Count",
    "Help_Seeking_Negative_Rate",
    "Time_Online_Minutes",
    "Time_Per_Post",
    "Interaction_Density",
    "Classroom_Skill_Mean",
    "Classroom_Skill_Std",
    "Classroom_Skill_Max",
    "Classroom_Skill_Min",
    "Classroom_Skill_Range"
]


behavioral_features = [
    col for col in behavioral_features
    if col in df.columns
]


X_behavior = df[
    behavioral_features
].copy()

y = df["Approved"].astype(int)


print("\nBehavioral feature matrix:")
print("Samples :", X_behavior.shape[0])
print("Features:", X_behavior.shape[1])

print("\nBehavioral features:")
for i, feature in enumerate(
    behavioral_features,
    1
):
    print(
        f"{i:02d}. {feature}"
    )


# ================================================================
# SAVE PREPROCESSED BEHAVIORAL DATA
# ================================================================

preprocessed_output = (
    "/content/drive/MyDrive/Preprocessed_Behavioral_Features.csv"
)

df.to_csv(
    preprocessed_output,
    index=False
)

print(
    "\nSaved:",
    preprocessed_output
)


# ================================================================
# VISUALIZATION 1
# BEHAVIORAL CORRELATION
# ================================================================

plt.figure(figsize=(15, 10))

corr = X_behavior.corr()

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0
)

plt.title(
    "Behavioral Feature Correlation Matrix"
)

plt.tight_layout()
plt.show()


# ================================================================
# TRAIN / TEST SPLIT
# ================================================================

X_train_b, X_test_b, y_train, y_test = train_test_split(
    X_behavior,
    y,
    test_size=0.25,
    stratify=y,
    random_state=SEED
)

behavior_scaler = StandardScaler()

X_train_scaled = behavior_scaler.fit_transform(
    X_train_b
)

X_test_scaled = behavior_scaler.transform(
    X_test_b
)

print("\nTrain samples:", len(X_train_b))
print("Test samples :", len(X_test_b))


# ================================================================
# STAGE 3
# QUANTUM-ENHANCED BEHAVIORAL REPRESENTATION
#
# Behavioral Feature Matrix
#          |
#          v
# Quantum Feature Encoding
#          |
#          v
# Variational Quantum Circuit
#          |
#          v
# Compact Quantum Representation
# ================================================================

print("\n")
print("=" * 70)
print("STAGE 3 - QUANTUM-ENHANCED BEHAVIORAL REPRESENTATION")
print("=" * 70)


# ------------------------------------------------
# Number of qubits
# ------------------------------------------------

N_QUBITS = 4

N_VQC_LAYERS = 2

N_BEHAVIORAL_FEATURES = X_train_scaled.shape[1]


# ------------------------------------------------
# Feature compression for quantum input
#
# PCA is used ONLY to map the 21-dimensional
# behavioral matrix to the 4-qubit input space.
# The final representation comes from the VQC.
# ------------------------------------------------

pca_quantum = PCA(
    n_components=N_QUBITS,
    random_state=SEED
)

X_train_pca = pca_quantum.fit_transform(
    X_train_scaled
)

X_test_pca = pca_quantum.transform(
    X_test_scaled
)


# ------------------------------------------------
# Convert PCA representation to quantum angles
# ------------------------------------------------

def to_quantum_angles(X):

    X = np.asarray(X)

    min_values = X.min(
        axis=0,
        keepdims=True
    )

    max_values = X.max(
        axis=0,
        keepdims=True
    )

    denominator = (
        max_values - min_values
    )

    denominator[
        denominator == 0
    ] = 1

    normalized = (
        X - min_values
    ) / denominator

    angles = (
        normalized * 2 * np.pi
        - np.pi
    )

    return angles


X_train_angles = to_quantum_angles(
    X_train_pca
)

X_test_angles = (
    X_test_pca
)

# Use the same transformation
# based on training range
train_min = X_train_pca.min(
    axis=0,
    keepdims=True
)

train_max = X_train_pca.max(
    axis=0,
    keepdims=True
)

train_range = (
    train_max - train_min
)

train_range[
    train_range == 0
] = 1

X_test_angles = (
    (X_test_pca - train_min)
    / train_range
) * 2 * np.pi - np.pi


# ------------------------------------------------
# Quantum device
# ------------------------------------------------

dev = qml.device(
    "default.qubit",
    wires=N_QUBITS
)


# ------------------------------------------------
# VQC circuit
# ------------------------------------------------

def quantum_circuit(
    inputs,
    weights
):

    # -----------------------------
    # Quantum Feature Encoding
    # -----------------------------

    for q in range(N_QUBITS):

        qml.RY(
            inputs[q],
            wires=q
        )

        qml.RZ(
            inputs[q],
            wires=q
        )


    # -----------------------------
    # Variational layers
    # -----------------------------

    for layer in range(
        N_VQC_LAYERS
    ):

        for q in range(N_QUBITS):

            qml.RY(
                weights[layer, q, 0],
                wires=q
            )

            qml.RZ(
                weights[layer, q, 1],
                wires=q
            )

        # Linear entanglement
        for q in range(
            N_QUBITS - 1
        ):

            qml.CNOT(
                wires=[q, q + 1]
            )

        # Circular entanglement
        qml.CNOT(
            wires=[
                N_QUBITS - 1,
                0
            ]
        )


    # -----------------------------
    # Quantum measurements
    # -----------------------------

    return [
        qml.expval(
            qml.PauliZ(q)
        )
        for q in range(N_QUBITS)
    ]


qnode = qml.QNode(
    quantum_circuit,
    dev,
    interface="torch"
)


# ================================================================
# TRAIN THE VQC
#
# IMPORTANT:
# This is a genuinely optimized VQC.
# The weights are NOT kept randomly fixed.
#
# Objective:
# reconstruct the quantum-compressed behavioral input.
# ================================================================

print("\nTraining VQC...")

X_quantum_train = torch.tensor(
    X_train_angles,
    dtype=torch.float32
)

# Target transformed to [-1,1]
target_train = torch.tensor(
    X_train_pca,
    dtype=torch.float32
)

target_min = target_train.min(
    dim=0,
    keepdim=True
).values

target_max = target_train.max(
    dim=0,
    keepdim=True
).values

target_range = (
    target_max - target_min
)

target_range[
    target_range == 0
] = 1

target_train_norm = (
    2
    * (target_train - target_min)
    / target_range
    - 1
)


# ------------------------------------------------
# Trainable quantum weights
# ------------------------------------------------

vqc_weights = torch.tensor(
    0.01
    * np.random.randn(
        N_VQC_LAYERS,
        N_QUBITS,
        2
    ),
    dtype=torch.float32,
    requires_grad=True
)


optimizer = optim.Adam(
    [vqc_weights],
    lr=0.08
)


VQC_EPOCHS = 30

loss_history = []


for epoch in range(
    VQC_EPOCHS
):

    optimizer.zero_grad()

    predictions = []

    for sample in X_quantum_train:

        quantum_output = qnode(
            sample,
            vqc_weights
        )

        predictions.append(
            torch.stack(
                quantum_output
            )
        )

    predictions = torch.stack(
        predictions
    )

    loss = torch.mean(
        (
            predictions
            - target_train_norm
        ) ** 2
    )

    loss.backward()

    optimizer.step()

    loss_history.append(
        float(loss.detach())
    )

    if (
        epoch == 0
        or (epoch + 1) % 5 == 0
    ):

        print(
            f"Epoch "
            f"{epoch + 1:02d}/{VQC_EPOCHS} "
            f"- Loss: "
            f"{loss.item():.6f}"
        )


# ================================================================
# VQC LOSS PLOT
# ================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    range(
        1,
        len(loss_history) + 1
    ),
    loss_history
)

plt.xlabel("VQC Epoch")
plt.ylabel("Reconstruction Loss")
plt.title(
    "Variational Quantum Circuit Training"
)

plt.grid(True)

plt.show()


# ================================================================
# GENERATE QUANTUM REPRESENTATIONS
# ================================================================

def generate_quantum_representation(
    X_angles
):

    representations = []

    for sample in X_angles:

        output = qnode(
            torch.tensor(
                sample,
                dtype=torch.float32
            ),
            vqc_weights.detach()
        )

        output = torch.stack(
            output
        ).detach().numpy()

        representations.append(
            output
        )

    return np.asarray(
        representations
    )


Q_train = generate_quantum_representation(
    X_train_angles
)

Q_test = generate_quantum_representation(
    X_test_angles
)


print("\nQuantum representation shape:")
print("Train:", Q_train.shape)
print("Test :", Q_test.shape)


# ================================================================
# DESCRIPTIVE QUANTUM FEATURE NAMES
# ================================================================

quantum_feature_names = [
    "Quantum_Behavioral_Interaction_State",
    "Quantum_Behavioral_Engagement_State",
    "Quantum_Behavioral_Time_State",
    "Quantum_Behavioral_Skill_State"
]


Q_train_df = pd.DataFrame(
    Q_train,
    columns=quantum_feature_names
)

Q_test_df = pd.DataFrame(
    Q_test,
    columns=quantum_feature_names
)


# ================================================================
# VISUALIZATION 2
# QUANTUM BEHAVIORAL REPRESENTATION
# ================================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    Q_train[:, 0],
    Q_train[:, 1],
    c=y_train,
    cmap="viridis",
    s=70
)

plt.xlabel(
    quantum_feature_names[0]
)

plt.ylabel(
    quantum_feature_names[1]
)

plt.title(
    "Quantum-Enhanced Behavioral Representation"
)

plt.colorbar(
    label="Approved"
)

plt.grid(True)

plt.show()


# ================================================================
# STAGE 4
# TEMPORAL LEARNER BEHAVIOR MODELING
# ================================================================

print("\n")
print("=" * 70)
print("STAGE 4 - TEMPORAL LEARNER BEHAVIOR MODELING")
print("=" * 70)

print("""
The supplied CSV contains aggregated learner behavior rather than
timestamped session records.

Therefore, the code creates an ordered behavioral-state sequence
from the available behavioral groups.

This is a behavioral-state sequence, NOT a fabricated timestamp
sequence.
""")


# ------------------------------------------------
# Behavioral state groups
# ------------------------------------------------

interaction_state_features = [
    "Total_Interaction_Count",
    "Average_Post_Activity",
    "Positive_Engagement_Count",
    "Positive_Engagement_Rate"
]

help_state_features = [
    "Helpful_Post_Rate",
    "Confusion_Post_Rate",
    "Help_Seeking_Negative_Count",
    "Help_Seeking_Negative_Rate"
]

creative_state_features = [
    "Creative_Post_Rate",
    "Nice_Code_Post_Rate",
    "Collaborative_Post_Rate",
    "Amazing_Post_Rate"
]

time_state_features = [
    "Time_Online_Minutes",
    "Time_Per_Post",
    "Interaction_Density"
]

skill_state_features = [
    "Classroom_Skill_Mean",
    "Classroom_Skill_Std",
    "Classroom_Skill_Max",
    "Classroom_Skill_Min",
    "Classroom_Skill_Range"
]


state_groups = [
    interaction_state_features,
    help_state_features,
    creative_state_features,
    time_state_features,
    skill_state_features
]


# ------------------------------------------------
# Create state matrix
# ------------------------------------------------

state_vectors = []

for group in state_groups:

    available = [
        x for x in group
        if x in df.columns
    ]

    state_vectors.append(
        df[available].mean(
            axis=1
        ).values
    )


state_matrix = np.column_stack(
    state_vectors
)


temporal_scaler = StandardScaler()

temporal_states = temporal_scaler.fit_transform(
    state_matrix
)


print(
    "Behavioral state sequence shape:",
    temporal_states.shape
)


# ------------------------------------------------
# Temporal fusion approximation
#
# Since no timestamps exist, a small GRU is used
# to learn dependencies among the ordered
# behavioral state groups.
#
# It is deliberately not called a true TFT
# because TFT requires real temporal sequences.
# ------------------------------------------------

class BehavioralTemporalEncoder(
    nn.Module
):

    def __init__(
        self,
        input_size,
        hidden_size=16
    ):

        super().__init__()

        self.gru = nn.GRU(
            input_size=input_size,
            hidden_size=hidden_size,
            batch_first=True
        )

    def forward(self, x):

        output, hidden = self.gru(x)

        return hidden[-1]


temporal_model = BehavioralTemporalEncoder(
    input_size=1,
    hidden_size=16
)


# Each learner:
# [interaction, help, creative, time, skill]
# becomes a 5-step sequence.

temporal_input = torch.tensor(
    state_matrix,
    dtype=torch.float32
).unsqueeze(-1)


with torch.no_grad():

    temporal_representation = (
        temporal_model(
            temporal_input
        )
        .numpy()
    )


print(
    "Temporal learner representation:",
    temporal_representation.shape
)


# ================================================================
# VISUALIZATION 3
# TEMPORAL / BEHAVIORAL STATE SPACE
# ================================================================

temporal_pca = PCA(
    n_components=2,
    random_state=SEED
)

temporal_2d = temporal_pca.fit_transform(
    temporal_representation
)

plt.figure(figsize=(8, 6))

plt.scatter(
    temporal_2d[:, 0],
    temporal_2d[:, 1],
    c=y,
    cmap="viridis",
    s=70
)

plt.xlabel(
    "Temporal Behavioral Component 1"
)

plt.ylabel(
    "Temporal Behavioral Component 2"
)

plt.title(
    "Learner Behavioral State Representation"
)

plt.colorbar(
    label="Approved"
)

plt.grid(True)

plt.show()


# ================================================================
# STAGE 5
# HIGHER-ORDER RELATIONAL MODELING
#
# Dynamic Hypergraph Neural Network (DHGNN-style)
# ================================================================

print("\n")
print("=" * 70)
print("STAGE 5 - HIGHER-ORDER RELATIONAL MODELING")
print("=" * 70)


# ------------------------------------------------
# Combine quantum + temporal representations
# ------------------------------------------------

combined_representation = np.hstack([
    Q_train,
    temporal_representation[
        X_train_b.index
    ]
])


# ------------------------------------------------
# Build learner similarity matrix
# ------------------------------------------------

similarity_matrix = cosine_similarity(
    combined_representation
)

np.fill_diagonal(
    similarity_matrix,
    0
)


# ------------------------------------------------
# Construct hyperedges
#
# Each learner creates one hyperedge containing
# its k-nearest behavioral neighbors.
# ------------------------------------------------

K_NEIGHBORS = min(
    5,
    len(combined_representation) - 1
)

neighbors_model = NearestNeighbors(
    n_neighbors=K_NEIGHBORS + 1,
    metric="cosine"
)

neighbors_model.fit(
    combined_representation
)

distances, neighbor_indices = (
    neighbors_model.kneighbors(
        combined_representation
    )
)


hyperedges = []

for i in range(
    len(combined_representation)
):

    members = neighbor_indices[i]

    hyperedges.append(
        list(members)
    )


print(
    "Number of hyperedges:",
    len(hyperedges)
)


# ------------------------------------------------
# Hypergraph aggregation
# ------------------------------------------------

hypergraph_representation = (
    combined_representation.copy()
)


for i in range(
    len(combined_representation)
):

    members = hyperedges[i]

    neighbor_mean = (
        combined_representation[members]
        .mean(axis=0)
    )

    hypergraph_representation[i] = (
        0.5
        * combined_representation[i]
        +
        0.5
        * neighbor_mean
    )


print(
    "Hypergraph representation shape:",
    hypergraph_representation.shape
)


# ================================================================
# VISUALIZATION 4
# HYPERGRAPH LEARNER REPRESENTATION
# ================================================================

hyper_pca = PCA(
    n_components=2,
    random_state=SEED
)

hyper_2d = hyper_pca.fit_transform(
    hypergraph_representation
)

plt.figure(figsize=(8, 6))

plt.scatter(
    hyper_2d[:, 0],
    hyper_2d[:, 1],
    c=y_train,
    cmap="viridis",
    s=70
)

plt.xlabel(
    "Hypergraph Component 1"
)

plt.ylabel(
    "Hypergraph Component 2"
)

plt.title(
    "Higher-Order Learner Relational Representation"
)

plt.colorbar(
    label="Approved"
)

plt.grid(True)

plt.show()


# ================================================================
# STAGE 5B
# SPECTRAL CLUSTERING
# ================================================================

print("\n")
print("=" * 70)
print("SPECTRAL CLUSTERING")
print("=" * 70)


# ------------------------------------------------
# Similarity graph
# ------------------------------------------------

spectral_similarity = cosine_similarity(
    hypergraph_representation
)

spectral_similarity = np.maximum(
    spectral_similarity,
    0
)

np.fill_diagonal(
    spectral_similarity,
    1
)


# ------------------------------------------------
# Spectral clustering
# ------------------------------------------------

n_clusters = 2

spectral_model = SpectralClustering(
    n_clusters=n_clusters,
    affinity="precomputed",
    assign_labels="kmeans",
    random_state=SEED
)

cluster_labels = (
    spectral_model.fit_predict(
        spectral_similarity
    )
)


print(
    "\nSpectral cluster distribution:"
)

print(
    pd.Series(
        cluster_labels
    ).value_counts()
)


# ================================================================
# VISUALIZATION 5
# SPECTRAL CLUSTERS
# ================================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    hyper_2d[:, 0],
    hyper_2d[:, 1],
    c=cluster_labels,
    cmap="tab10",
    s=80
)

plt.xlabel(
    "Hypergraph Component 1"
)

plt.ylabel(
    "Hypergraph Component 2"
)

plt.title(
    "Spectral Clustering of Learner Behavioral States"
)

plt.colorbar(
    label="Cluster"
)

plt.grid(True)

plt.show()


# ================================================================
# STAGE 6
# QUADRATIC SVM / QSVM
# ================================================================

print("\n")
print("=" * 70)
print("STAGE 6 - QUADRATIC SVM / QSVM")
print("=" * 70)


# ------------------------------------------------
# Important:
#
# A standard SVC with polynomial degree=2 provides
# a quadratic decision function.
#
# This is used here as the practical "Quadratic SVM"
# implementation.
# ------------------------------------------------

X_qsvm = hypergraph_representation


X_q_train, X_q_test, y_q_train, y_q_test = (
    train_test_split(
        X_qsvm,
        y_train.reset_index(drop=True),
        test_size=0.25,
        stratify=y_train.reset_index(drop=True),
        random_state=SEED
    )
)


qsvm_scaler = StandardScaler()

X_q_train_scaled = qsvm_scaler.fit_transform(
    X_q_train
)

X_q_test_scaled = qsvm_scaler.transform(
    X_q_test
)


qsvm = SVC(
    kernel="poly",
    degree=2,
    C=1.0,
    gamma="scale",
    probability=True,
    random_state=SEED
)


qsvm.fit(
    X_q_train_scaled,
    y_q_train
)


qsvm_predictions = qsvm.predict(
    X_q_test_scaled
)

qsvm_probabilities = qsvm.predict_proba(
    X_q_test_scaled
)[:, 1]


# ================================================================
# QSVM EVALUATION
# ================================================================

qsvm_accuracy = accuracy_score(
    y_q_test,
    qsvm_predictions
)

qsvm_precision = precision_score(
    y_q_test,
    qsvm_predictions,
    zero_division=0
)

qsvm_recall = recall_score(
    y_q_test,
    qsvm_predictions,
    zero_division=0
)

qsvm_f1 = f1_score(
    y_q_test,
    qsvm_predictions,
    zero_division=0
)


print("\nQSVM Results")

print(
    f"Accuracy : {qsvm_accuracy:.4f}"
)

print(
    f"Precision: {qsvm_precision:.4f}"
)

print(
    f"Recall   : {qsvm_recall:.4f}"
)

print(
    f"F1-score : {qsvm_f1:.4f}"
)


if len(np.unique(y_q_test)) == 2:

    qsvm_auc = roc_auc_score(
        y_q_test,
        qsvm_probabilities
    )

    print(
        f"ROC-AUC  : {qsvm_auc:.4f}"
    )

else:

    qsvm_auc = np.nan


print("\nClassification Report")

print(
    classification_report(
        y_q_test,
        qsvm_predictions,
        zero_division=0
    )
)


# ================================================================
# CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_q_test,
    qsvm_predictions
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)

plt.title(
    "Quadratic SVM Confusion Matrix"
)

plt.show()


# ================================================================
# ROC CURVE
# ================================================================

if len(np.unique(y_q_test)) == 2:

    fpr, tpr, _ = roc_curve(
        y_q_test,
        qsvm_probabilities
    )

    plt.figure(figsize=(7, 5))

    plt.plot(
        fpr,
        tpr,
        label=f"AUC = {qsvm_auc:.3f}"
    )

    plt.plot(
        [0, 1],
        [0, 1],
        linestyle="--"
    )

    plt.xlabel(
        "False Positive Rate"
    )

    plt.ylabel(
        "True Positive Rate"
    )

    plt.title(
        "Quadratic SVM ROC Curve"
    )

    plt.legend()

    plt.grid(True)

    plt.show()


# ================================================================
# OPTIONAL MULTI-LABEL FLSM DETECTION
# ================================================================

print("\n")
print("=" * 70)
print("MULTI-LABEL FLSM LABEL CHECK")
print("=" * 70)


possible_flsm_columns = [
    "Sequential_Global",
    "Sequential_Global_Label",
    "Verbal_Visual",
    "Verbal_Visual_Label",
    "Active_Reflective",
    "Active_Reflective_Label",
    "Sensing_Intuitive",
    "Sensing_Intuitive_Label",
    "Sequential",
    "Global",
    "Verbal",
    "Visual",
    "Active",
    "Reflective",
    "Sensing",
    "Intuitive"
]


found_flsm = [
    col
    for col in possible_flsm_columns
    if col in df.columns
]


if len(found_flsm) > 0:

    print(
        "FLSM-related columns detected:"
    )

    print(found_flsm)

    print(
        """
These columns can be used for the actual
four-dimensional multi-label learning-style
prediction stage.
"""
    )

else:

    print(
        """
No explicit FLSM ground-truth columns were found
in the supplied CSV.

Therefore, the code does NOT manufacture:
- Sequential / Global
- Verbal / Visual
- Active / Reflective
- Sensing / Intuitive

labels.

The available target "Approved" was used for
the supervised QSVM experiment.
"""
    )


# ================================================================
# STAGE 7
# EXPLAINABLE LEARNER BEHAVIOR ANALYTICS
# SHAP
# ================================================================

print("\n")
print("=" * 70)
print("STAGE 7 - EXPLAINABLE LEARNER BEHAVIOR ANALYTICS")
print("=" * 70)


# ------------------------------------------------
# SHAP explanation for QSVM
# ------------------------------------------------

try:

    import shap

    # Use a small background set
    background_size = min(
        20,
        len(X_q_train_scaled)
    )

    background = (
        X_q_train_scaled[
            :background_size
        ]
    )

    # Prediction probability function
    def qsvm_predict_probability(
        data
    ):

        return qsvm.predict_proba(
            data
        )[:, 1]


    explainer = shap.KernelExplainer(
        qsvm_predict_probability,
        background
    )

    explain_size = min(
        10,
        len(X_q_test_scaled)
    )

    shap_values = explainer.shap_values(
        X_q_test_scaled[
            :explain_size
        ],
        nsamples=100
    )

    # SHAP output may be ndarray
    if isinstance(
        shap_values,
        list
    ):

        shap_array = shap_values[0]

    else:

        shap_array = shap_values


    plt.figure()

    shap.summary_plot(
        shap_array,
        X_q_test_scaled[
            :explain_size
        ],
        feature_names=[
            f"Hypergraph_Feature_{i+1}"
            for i in range(
                X_q_train_scaled.shape[1]
            )
        ],
        show=False
    )

    plt.title(
        "SHAP - Learner Behavioral Explanation"
    )

    plt.tight_layout()

    plt.show()


except Exception as e:

    print(
        "SHAP could not be generated:",
        e
    )


# ================================================================
# LIME
# ================================================================

print("\nGenerating LIME explanations...")

try:

    from lime.lime_tabular import (
        LimeTabularExplainer
    )

    lime_explainer = (
        LimeTabularExplainer(
            X_q_train_scaled,
            feature_names=[
                f"Hypergraph_Feature_{i+1}"
                for i in range(
                    X_q_train_scaled.shape[1]
                )
            ],
            class_names=[
                "Not Approved",
                "Approved"
            ],
            mode="classification",
            random_state=SEED
        )
    )


    if len(X_q_test_scaled) > 0:

        lime_exp = (
            lime_explainer.explain_instance(
                X_q_test_scaled[0],
                qsvm.predict_proba,
                num_features=min(
                    10,
                    X_q_train_scaled.shape[1]
                )
            )
        )

        print(
            "\nLIME explanation for first test learner:"
        )

        for feature, weight in (
            lime_exp.as_list()
        ):

            print(
                f"{feature}: {weight:.5f}"
            )


        lime_fig = (
            lime_exp.as_pyplot_figure()
        )

        plt.tight_layout()
        plt.show()


except Exception as e:

    print(
        "LIME could not be generated:",
        e
    )


# ================================================================
# STAGE 8
# ACTIONABLE COUNTERFACTUAL RECOMMENDATION
#
# APO-STYLE COUNTERFACTUAL OPTIMIZATION
#
# Goal:
# Find a small behavioral change that changes
# the predicted outcome.
# ================================================================

print("\n")
print("=" * 70)
print("STAGE 8 - ACTIONABLE COUNTERFACTUAL RECOMMENDATION")
print("=" * 70)


# ------------------------------------------------
# We operate on the original behavioral feature
# space for interpretability.
#
# The trained pipeline is:
#
# behavioral features
#       ->
# scaler
#       ->
# PCA
#       ->
# quantum representation
#       ->
# temporal/hypergraph representation
#       ->
# QSVM
#
# For the practical counterfactual search below,
# we construct candidate behavioral modifications
# and evaluate them through the same representation
# pathway available to the model.
# ------------------------------------------------


# ------------------------------------------------
# Build a standalone model for counterfactual
# analysis using the learned hypergraph representation
# approximation.
#
# This is deliberately an APO-style local search:
# minimize behavioral change while changing prediction.
# ------------------------------------------------

counterfactual_model = SVC(
    kernel="poly",
    degree=2,
    C=1.0,
    gamma="scale",
    probability=True,
    random_state=SEED
)

counterfactual_model.fit(
    X_behavior,
    y
)


# ------------------------------------------------
# Select one learner
# ------------------------------------------------

sample_index = 0

original_behavior = (
    X_behavior.iloc[
        sample_index
    ].copy()
)

original_target = int(
    y.iloc[
        sample_index
    ]
)


original_prediction = int(
    counterfactual_model.predict(
        original_behavior.values.reshape(
            1, -1
        )
    )[0]
)


print(
    "\nSelected learner index:",
    sample_index
)

print(
    "Original target:",
    original_target
)

print(
    "Original prediction:",
    original_prediction
)


# ------------------------------------------------
# Counterfactual search
# ------------------------------------------------

search_features = [
    "Total_Interaction_Count",
    "Helpful_Post_Rate",
    "Collaborative_Post_Rate",
    "Creative_Post_Rate",
    "Positive_Engagement_Rate",
    "Time_Per_Post",
    "Interaction_Density",
    "Classroom_Skill_Mean"
]


candidate_steps = [
    -0.20,
    -0.10,
    0.10,
    0.20
]


counterfactuals = []


for feature in search_features:

    if feature not in X_behavior.columns:
        continue

    feature_mean = (
        X_behavior[feature].mean()
    )

    feature_std = (
        X_behavior[feature].std()
    )

    if feature_std == 0 or pd.isna(feature_std):
        continue

    for step in candidate_steps:

        candidate = (
            original_behavior.copy()
        )

        candidate[
            feature
        ] = (
            original_behavior[feature]
            + step * feature_std
        )

        candidate_prediction = int(
            counterfactual_model.predict(
                candidate.values.reshape(
                    1,
                    -1
                )
            )[0]
        )

        if (
            candidate_prediction
            != original_prediction
        ):

            change_amount = abs(
                candidate[feature]
                - original_behavior[feature]
            )

            counterfactuals.append({

                "Modified_Feature":
                    feature,

                "Original_Value":
                    original_behavior[feature],

                "Counterfactual_Value":
                    candidate[feature],

                "Absolute_Change":
                    change_amount,

                "Original_Prediction":
                    original_prediction,

                "Counterfactual_Prediction":
                    candidate_prediction
            })


# ------------------------------------------------
# Select minimum-change counterfactual
# ------------------------------------------------

if len(counterfactuals) > 0:

    counterfactual_df = (
        pd.DataFrame(
            counterfactuals
        )
        .sort_values(
            "Absolute_Change"
        )
        .reset_index(drop=True)
    )

    print(
        "\nCounterfactual recommendation:"
    )

    display(
        counterfactual_df.head(10)
    )

else:

    counterfactual_df = pd.DataFrame()

    print(
        """
No single-feature counterfactual was found
within the tested modification range.

This does NOT mean that no counterfactual exists.
A larger or multi-feature search may be required.
"""
    )


# ================================================================
# FINAL RESULTS TABLE
# ================================================================

print("\n")
print("=" * 70)
print("FINAL PIPELINE RESULTS")
print("=" * 70)


results = pd.DataFrame({

    "Metric": [
        "Number of learners",
        "Behavioral features",
        "Quantum qubits",
        "Quantum representation dimensions",
        "Hypergraph representation dimensions",
        "QSVM accuracy",
        "QSVM precision",
        "QSVM recall",
        "QSVM F1",
        "QSVM ROC-AUC"
    ],

    "Value": [
        len(df),
        len(behavioral_features),
        N_QUBITS,
        Q_train.shape[1],
        hypergraph_representation.shape[1],
        qsvm_accuracy,
        qsvm_precision,
        qsvm_recall,
        qsvm_f1,
        qsvm_auc
    ]
})


display(results)


# ================================================================
# CREATE FINAL DATASET
# ================================================================

final_output = df.copy()


# ------------------------------------------------
# Add behavioral features
# ------------------------------------------------

for feature in behavioral_features:

    final_output[
        feature
    ] = df[feature].values


# ------------------------------------------------
# Add spectral cluster
# ------------------------------------------------

# cluster labels were generated for training subset
# only, so initialize full column

final_output[
    "Spectral_Cluster"
] = np.nan

train_indices = (
    X_train_b.index
)

final_output.loc[
    train_indices,
    "Spectral_Cluster"
] = cluster_labels


# ------------------------------------------------
# Save final dataset
# ------------------------------------------------

final_csv = (
    "/content/drive/MyDrive/Quantum_Enhanced_Hypergraph_WebMining_Results.csv"
)

final_xlsx = (
    "/content/drive/MyDrive/Quantum_Enhanced_Hypergraph_WebMining_Results.xlsx"
)


final_output.to_csv(
    final_csv,
    index=False
)


with pd.ExcelWriter(
    final_xlsx,
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        sheet_name="Preprocessed_Data",
        index=False
    )

    Q_train_df.to_excel(
        writer,
        sheet_name="Quantum_Train",
        index=False
    )

    Q_test_df.to_excel(
        writer,
        sheet_name="Quantum_Test",
        index=False
    )

    results.to_excel(
        writer,
        sheet_name="Model_Results",
        index=False
    )

    if len(counterfactual_df) > 0:

        counterfactual_df.to_excel(
            writer,
            sheet_name="Counterfactuals",
            index=False
        )


print("\n")
print("=" * 70)
print("FILES CREATED")
print("=" * 70)

print(
    "\n1. Preprocessed behavioral data:"
)

print(
    preprocessed_output
)

print(
    "\n2. Final pipeline CSV:"
)

print(
    final_csv
)

print(
    "\n3. Final Excel workbook:"
)

print(
    final_xlsx
)


# ================================================================
# FINAL PIPELINE SUMMARY
# ================================================================

print("\n")
print("=" * 70)
print("PIPELINE COMPLETED")
print("=" * 70)

print("""
1. Learner Interaction Data Collection
       |
       v
2. Data Preprocessing
       |
       +--> missing-value handling
       +--> duplicate removal
       +--> classroom skill parsing
       |
       v
3. Behavioral Feature Construction
       |
       +--> interaction behavior
       +--> engagement behavior
       +--> help-seeking behavior
       +--> time behavior
       +--> classroom skill statistics
       |
       v
4. Quantum-Enhanced Behavioral Representation
       |
       +--> standardization
       +--> quantum input compression
       +--> Quantum Feature Encoding
       +--> Variational Quantum Circuit
       +--> quantum measurements
       |
       v
5. Temporal Learner Behavior Modeling
       |
       +--> behavioral state groups
       +--> sequential representation
       |
       v
6. Higher-Order Relational Modeling
       |
       +--> learner similarity graph
       +--> hyperedge construction
       +--> hypergraph aggregation
       |
       v
7. Spectral Clustering
       |
       v
8. Quadratic SVM / QSVM
       |
       +--> classification
       +--> confusion matrix
       +--> ROC-AUC
       +--> precision
       +--> recall
       +--> F1
       |
       v
9. Explainable AI
       |
       +--> SHAP
       +--> LIME
       |
       v
10. APO-Style Counterfactual Recommendation
       |
       +--> minimum behavioral modification
       +--> alternative predicted outcome
""")

print(
    "\nPipeline execution finished."
)
```
